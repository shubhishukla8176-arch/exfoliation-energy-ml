```python
# Import the JARVIS data retrieval function

from jarvis.db.figshare import data

print("Connecting to JARVIS...")

# Retrieve the 3D materials database
dft_3d = data("dft_3d")

print("Database retrieved successfully!")

# Check the number of records
print("Total records:", len(dft_3d))
```

```python
# Look at the first material record

print(dft_3d[0])
```

```python
# Display all available property names

print(dft_3d[0].keys())
```

```python
import pandas as pd

# Convert JARVIS records into a DataFrame
df = pd.DataFrame(dft_3d)

# Display the first five rows
display(df.head())
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Convert the target into numerical values
df["exfoliation_energy"] = pd.to_numeric(
    df["exfoliation_energy"],
    errors="coerce"
)

# Remove records without exfoliation energy
df_clean = df.dropna(
    subset=["exfoliation_energy"]
).copy()

print("Total materials:", len(df))
print("Materials with exfoliation energy:", len(df_clean))
```

```python
import os

project_folder = "/content/drive/MyDrive/MASC520_Project"

os.makedirs(project_folder, exist_ok=True)

print("Project folder created!")
```

```python
# Select the essential information
required_columns = [
    "jid",
    "atoms",
    "exfoliation_energy"
]

df_final = df_clean[required_columns].copy()

# Save structures and target values in JSON
json_path = os.path.join(
    project_folder,
    "exfoliation_energy_dataset.json"
)

df_final.to_json(
    json_path,
    orient="records",
    indent=2
)

# Save a separate CSV without nested crystal structures
csv_path = os.path.join(
    project_folder,
    "exfoliation_energy_metadata.csv"
)

df_final.drop(columns=["atoms"]).to_csv(
    csv_path,
    index=False
)

print("Both files saved successfully!")
```
